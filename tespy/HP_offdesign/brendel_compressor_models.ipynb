# Reciprocating compressor correlations (Brendel et al.)

**References**

- Doughty, Brendel, Bernal, Arpagaus, Bertsch (2024): *Set of Performance
  Correlations for Reciprocating Compressors Covering Synthetic and
  Hydrocarbon Refrigerants*. 27th International Compressor Engineering
  Conference at Purdue, paper 1313. https://docs.lib.purdue.edu/icec/
- Brendel, Bernal, Luechinger, Arpagaus, Bertsch (2026): *Experimental data
  of a reciprocating compressor across wide range of refrigerants and
  operating conditions*. Int. J. Refrigeration 182, 409-417.
  https://doi.org/10.1016/j.ijrefrig.2025.11.012


```python
# London correlation (Brendel et al. 2023), coefficients from Table 5 of
# Brendel et al. (2026); Purdue correlation (Doughty et al. 2024),
# coefficients from Table 6
LONDON = {
    "a0": 0.66981, "a1": 0.01466, "a2": 0.00838,  # a2 in 1/kPa
    "a3": 0.00102,
    "b0": 0.08244, "b1": 0.72773,
}
PURDUE = {
    "c0": 0.6462, "c1": 0.5798, "c2": 0.0077,     # c2 in 1/kPa
    "c3": 0.0012, "c4": 0.0012,                   # c4 in 1/K
    "R_syn": 1.712, "R_HC": 2.047,
    "b0": 0.0824, "b1": 0.7277,                   # rounded London values
}


def eta_ois_london(p_suc, pr):
    """London isentropic efficiency, p_suc in kPa, dT_sh in K"""
    c = LONDON
    return c["a0"] - 0.6 / (pr - c["a1"]) ** (c["a2"] * p_suc) \
        - c["a3"] * pr ** 1.8


def eta_vol_london(pr):
    """London volumetric efficiency"""
    c = LONDON
    return 1 - c["b0"] * (pr - 1) ** c["b1"]


def eta_ois_purdue(p_suc, pr, dT_sh, hydrocarbon=False):
    """purdue isentropic efficiency, p_suc in kPa, dT_sh in K"""
    c = PURDUE
    R = c["R_HC"] if hydrocarbon else c["R_syn"]
    return c["c0"] - c["c1"] / pr ** (c["c2"] * p_suc) \
        - c["c3"] * pr ** R + c["c4"] * dT_sh


def eta_vol_purdue(pr):
    """Volumetric efficiency by the Purdue paper (same as London)"""
    c = PURDUE
    return 1 - c["b0"] * (pr - 1) ** c["b1"]
```
